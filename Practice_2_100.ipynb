```python
# 98 np.einsum for inner outer sum and multiplication

import numpy as np

# Create two vectors
A = np.array([1, 2, 3])

B = np.array([4, 5, 6])

print("Vector A:", A)

print("Vector B:", B)

# 1. INNER PRODUCT
inner = np.einsum('i,i->', A, B)

# 'i,i->'
# Multiply matching elements
# Then sum them

# Equivalent to:
# np.inner(A,B)
# or np.dot(A,B)

print("\nInner Product:")

print(inner)

# Manual:
# (1*4) + (2*5) + (3*6)

outer = np.einsum('i,j->ij', A, B)

# 'i,j->ij'
# Every element of A multiplies
# every element of B

# Result shape:
# (3,1) × (1,3) → (3,3)

print("\nOuter Product:")

print(outer)
mul = np.einsum('i,i->i', A, B)

# 'i,i->i'
# Multiply matching elements
# Keep result as vector

print("\nElement-wise Multiplication:")

print(mul)

# Equivalent to:
# A * B

sum_A = np.einsum('i->', A)

# 'i->'
# Reduce all elements into single scalar

print("\nSum of A:")

print(sum_A)

# Equivalent to:
# np.sum(A)
```

```python
# sampling along with path 
import numpy as np

# Create path coordinates

X = np.array([0, 1, 2, 4, 7])

Y = np.array([0, 1, 1, 3, 4])

print("X Coordinates:", X)

print("Y Coordinates:", Y)

dx = np.diff(X)

dy = np.diff(Y)

# diff()
# Computes difference between neighboring values

print("\ndx:", dx)

print("dy:", dy)


dist = np.sqrt(dx**2 + dy**2)

# Euclidean distance formula

print("\nSegment Distances:")

print(dist)
# STEP 2
# Compute cumulative distance

cumdist = np.insert(np.cumsum(dist), 0, 0)

# cumsum()
# Running cumulative sum

# insert(...,0,0)
# Add starting distance 0

print("\nCumulative Distance:")

print(cumdist)
# STEP 3
# Generate equally spaced distances

num_samples = 10

target_dist = np.linspace(0, cumdist[-1], num_samples)

# linspace(start,end,n)
# Creates equally spaced values

print("\nTarget Distances:")

print(target_dist)
# STEP 4
# Interpolate X and Y coordinates

X_sampled = np.interp(target_dist, cumdist, X)

Y_sampled = np.interp(target_dist, cumdist, Y)

# interp()
# Interpolates values at new positions

print("\nSampled X Coordinates:")

print(X_sampled)

print("\nSampled Y Coordinates:")

print(Y_sampled)

# Final sampled path

sampled_points = np.column_stack((X_sampled, Y_sampled))

# column_stack()
# Combines vectors as columns

print("\nEquidistant Sampled Points:")

print(sampled_points)
```

```python
# multidiemntional draw row selection

import numpy as np

# Given integer n
n = 5

# Create 2D array
X = np.array([
    [2, 1, 2],     # valid
    [1, 1, 1],     # sum != 5
    [3, 0, 2],     # valid
    [1.5, 2, 1.5], # not integers
    [5, 0, 0],     # valid
    [2, 2, 2]      # sum = 6
])

print("Original Array:\n", X)
# STEP 1
# Check integer condition

integer_mask = (np.mod(X, 1) == 0)

# mod(X,1)
# Fractional remainder after division by 1

# Integer values produce remainder 0

print("\nInteger Mask:\n", integer_mask)

all_integer = np.logical_and.reduce(integer_mask, axis=1)

# logical_and.reduce()
# Checks if all values in each row are True

print("\nRows With All Integers:")

print(all_integer)
# STEP 2
# Check row sums equal n

sum_mask = (X.sum(axis=1) == n)

print("\nRows Whose Sum Equals n:")

print(sum_mask)
# STEP 3
# Combine both conditions

final_mask = all_integer & sum_mask

# Both conditions must be True

print("\nFinal Valid Row Mask:")

print(final_mask)

# STEP 4
# Select valid rows


valid_rows = X[final_mask]

print("\nRows Interpreted As Multinomial Draws:")

print(valid_rows)
```

```python
# 79 distance from points to lines

P0 = np.array([ [0, 0],[1, 1]])

P1 = np.array([ [1, 0],[2, 2]])

P = np.array([[1, 1],[2, 0],[0, 2]])
# Direction vector of each line
# (line end - line start)
line_vectors = P1 - P0

# Create vector from line start to each point
point_vectors = P[:, None, :] - P0            #here (3,2) becomes (3,1,2) by [:,none,:] that is point line coordinates , every poitns can be substarct from every line
# Cross product magnitude
# Formula:
# |(x1*y2 - y1*x2)|
cross_product = np.abs( line_vectors[:, 0] * point_vectors[:, :, 1]- line_vectors[:, 1] * point_vectors[:, :, 0])

# line 17 extracting  x and y coordinates,, here tranpose needed for shape alignment
# Length of each line vector
line_length = np.linalg.norm(line_vectors,axis=1)
# Distance formula
# Distance = Area / Base
distance = cross_product / line_length

print(distance)

# line vector computation [1,0] - [0,0]. then will become [1,0] and [2,2]- [1,1] =[1,1]
# line direction vector is needed for distance computation
# in numpy
# for compute vision - detects distance from objects to edges, shape machine
# gis /map - distance location from roads
# robot path planning and obstacle avoidance
# CAD = point to line measurement

# machine learning = geomatric feature engineering and spatial clustering
```

```python
# 80 extract fixed size subarray 

def extract_subarray(arr, center, shape, fill_value=0):

    result = np.full(shape, fill_value)

    r, c = center
    h, w = shape

    r_start = max(r - h // 2, 0)
    r_end   = min(r + h // 2 + 1, arr.shape[0])

    c_start = max(c - w // 2, 0)
    c_end   = min(c + w // 2 + 1, arr.shape[1])

    res_r_start = max(0, h // 2 - r)
    res_c_start = max(0, w // 2 - c)

    result[
        res_r_start:res_r_start + (r_end-r_start),
        res_c_start:res_c_start + (c_end-c_start)
    ] = arr[r_start:r_end, c_start:c_end]

    return result


A = np.arange(25).reshape(5,5)

print(extract_subarray(A,(0,0),(3,3),fill_value=-1))

# line 5 np.full(shape, value) Creates an array filled with a constant value like np.full((2,2)9) = [9,9][9,9]
# uses iin image processing convolutional windows medical imaging object detection
```

```python
# conveys game of life using numpy

def game_of_life(Z):

    # Count neighbors
    N = (
        np.roll(Z, 1, 0) + np.roll(Z,-1,0) +
        np.roll(Z, 1, 1) + np.roll(Z,-1,1) +
        np.roll(np.roll(Z,1,0),1,1) +
        np.roll(np.roll(Z,1,0),-1,1) +
        np.roll(np.roll(Z,-1,0),1,1) +
        np.roll(np.roll(Z,-1,0),-1,1)
    )

    # Apply rules
    birth = (N == 3) & (Z == 0)

    survive = ((N == 2) | (N == 3)) & (Z == 1)

    return birth | survive


Z = np.random.randint(0,2,(5,5))

print("Before:")
print(Z)

print("After:")
print(game_of_life(Z).astype(int))

# line 7,8 np.roll () that is shifts array like A=np.array[1,2,3,4] np.roll(A,1)=[4,1,2,3] shift by one placce

# use in simulations , artificial life models , scientific research ,cellular automata

```

```python
# 91 record array from regular array 

# normal numpy to record array [1,2][3,4][5,6] into record array x= 1 3 5 , and y = 2 4 6

import numpy as np

# Regular array
Z = np.array([[1,2],[3,4],[5,6]])    # here x= 1,3,5. and y=2,4,6

# Create record array
R = np.core.records.fromarrays(Z.T,names='x,y')

print(R)

# fromarrays expects column not rows

# what is record array z=[:,0] can write R.x (more readable

#Used in:

# Financial datasets
# Scientific measurements
#Structured records
#financial dtasets

```
