```python
#Create arr = np.array([300, 150, 50], dtype=np.int32). Cast to int8. What is the result and why?

arr = np.array([300, 150, 50], dtype=np.int32)
arr_int8 = arr.astype(np.int8) # When we cast from int32 to int8,wrapped around due to overflow. The int8 data type can only store values from -128 to 127. So, 300 becomes 44 (300 - 256), 150 becomes -106 (150 - 256), and 50 remains 50 since it is within the int8 range. #256 is the range of values that int8 can store, so when we exceed that range, it wraps around. #256 minus 256 is the same as adding -256, so it wraps around to the correct value within the int8 range.
print(arr_int8) 
```

```python
# temperature reading over 30 days

temps = np.array([22,35,28,41,19,37,30,25,33,18,
                  40,29,23,36,31,27,38,20,24,34,
                  21,39,26,32,17,43,28,35,22,30])
print(temps)
```

```python
#temperature was above 35°C (heat wave).
heat = temps[temps > 35]
print(heat)
```

```python
#temp below 20°C (cold day).
cold = temps[temps < 20]
print(cold)
```

```python
#temperature was between 25°C and 35°C (comfortable0

comfortable = temps[(temps >= 25) & (temps <= 35)]
print(comfortable)  
```

```python
#get how many days were comfortable
comfortable_days = comfortable.size #
print(comfortable_days) 
```

```python
#either heat wave, cold day(extreme days)

extreme_days = np.concatenate((heat, cold)) # The np.concatenate() function is used to combine the heat and cold arrays into a single array called extreme_days. This array will contain all the temperatures that were either above 35°C (heat wave) or below 20°C (cold day).
print(extreme_days)
```

```python
#were   all days  above 15 degrees?

above_15 = np.all(temps > 15) # The np.all() function checks if all elements in the temps array are greater than 15. If all temperatures are above 15°C, it will return True; otherwise, it will return False.
print(above_15) 
```

```python
# any days above 42 degrees?temp ? use np.any() function to check if any temperature in the temps array is above 42 degrees. The np.any() function will return True if at least one element in the array satisfies the condition (greater than 42), and False otherwise.
above_42 = np.any(temps > 42)
print(above_42)
```

```python
#NaN and inf handling (Data cleaning)

#given sensor data from an iot devuce

sensor = np.array([23.1,np.nan,24.5, np.inf, 22.8,np.nan,25.3,-np.inf,21.9,24.0,np.nan,23.7])
print(sensor)
```

```python
# count number of Nan values

nan_count = np.isnan(sensor).sum() # The np.isnan() function returns a boolean array where True indicates the presence of NaN values. By summing this boolean array, we can count the total number of NaN values in the sensor array.
print(nan_count)
```

```python
#count  inf both +ve and -ve

inf_count = np.isinf(sensor).sum() # The np.isinf() function returns a boolean array where True indicates the presence of infinite values (both positive and negative). By summing this boolean array, we can count the total number of infinite values in the sensor array.
print(inf_count)    
```

```python
#extract valid (finite) readings

valid_readings = sensor[np.isfinite(sensor)] # The np.isfinite() function returns a boolean array where True indicates the presence of finite values (not NaN or inf). By using this boolean array to index the sensor array, we can extract only the valid (finite) readings.
print(valid_readings)   
```

```python
#replace NaN values with mean of all finite values

mean_finite = np.nanmean(sensor) 
```

```python
#Replace Inf values with the max finite value and -Inf with the min finite 

max_finite = np.nanmax(sensor)
min_finite = np.nanmin(sensor)  

sensor[np.isinf(sensor)] = np.where(np.isposinf(sensor), max_finite, min_finite)  
```

```python
#calculate nanmean, nanstd, nanmin, nanmax of the sensor array
nanmean = np.nanmean(sensor)
nanstd = np.nanstd(sensor)
nanmin = np.nanmin(sensor)
nanmax = np.nanmax(sensor)  
print(nanmean)
print( nanstd)        
print(nanmin)
print( nanmax)
```

```python
# sales dataset

sales = np.array([
    [1500, 1800, np.nan, 2100, 1950, 2300, np.nan, 1700, 2000, 2150, 1600, 1900],
    [800,  np.nan, 950, 1100, 850,  np.inf, 900,  1050, np.nan, 800,  920,  1000],
    [3000, 3200, 2900, np.nan, 3100, 3500, 3300, np.nan, 3050, 3400, 3150, 3600],
    [450,  500,  np.nan, 600,  -100, 550,  np.nan, 480,  520,  -np.inf, 490, 530],
    [2200, 2100, 2400, 2300, np.nan, 2500, 2350, 2450, 2100, 2300, np.nan, 2600],
    [100,  120,  90,   np.nan, 130, 110,  140,   np.nan, 100, 120,  np.inf, 115]
])
print(sales)
```

```python
#shape of dataset
print(sales.shape)  
```

```python
#calculate nan inf and negative values in datasets

nan_count = np.isnan(sales).sum()
inf_count = np.isinf(sales).sum()
negative_count = np.sum(sales < 0) 
print( nan_count)
print( inf_count)
print( negative_count)
```

```python
#Clean the data replace NaN with row-wise nanmean, replace Inf/-Inf/negatives with 0
row_nanmean = np.nanmean(sales, axis=1)
for i in range(sales.shape[0]):
    sales[i, np.isnan(sales[i])] = row_nanmean[i]
sales[np.isinf(sales) | (sales < 0)] = 0
print(sales)           





```

```python
#Find which month (column index) had the highest total sales across all products
total_sales_per_month = np.nansum(sales, axis=0) # The np.nansum() function is used to calculate the sum of sales for each month while ignoring any NaN values. By specifying axis=0, we are summing across the rows (products) for each column (month).
highest_sales_month = np.argmax(total_sales_per_month) # The np.argmax() function is used to find the index of the month with the highest total sales. It returns the index of the maximum value in the total_sales_per_month array, which corresponds to the month with the highest sales.
print(highest_sales_month)      
```

```python
#which month had total sales above oveerall mean sales?
overall_mean_sales = np.nanmean(sales)
months_above_mean = np.where(total_sales_per_month > overall_mean_sales)[0] #[0] is used to extract the indices from the result of np.where(), which returns a tuple of arrays. By specifying [0], we are accessing the first array in the tuple, which contains the indices of the months where total sales are above the overall mean sales.      
print(months_above_mean)    
```

```python
#Cast the cleaned array to float32 to save memory. How much memory was saved?

sales_float32 = sales.astype(np.float32)
memory_float32 = sales_float32.nbytes / (1024 * 1024) # Convert bytes to megabytes
print(memory_float32)    #memory was saved by casting to float32. The original sales array was likely of a higher precision data type (like float64), so by casting it to float32, we reduce the memory usage. The exact amount of memory saved can be calculated by comparing the nbytes of the original array with that of the float32 array.
```
