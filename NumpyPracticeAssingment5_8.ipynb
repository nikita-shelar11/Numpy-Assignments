Find:

Temperatures above 38.0

Count normal readings (36.0–37.5)

Percentage of high fever readings (>39.0)

````python
import numpy as np

temperatures = np.array([36.5, 37.2, 38.9, 36.1,
                         39.5, 37.8, 40.1, 36.8])
# Above 38°C

fever = temperatures[temperatures > 38.0]

print(fever)
# Normal temperatures
normal = temperatures[(temperatures >= 36.0) &
                      (temperatures <= 37.5)]

# High fever (>39°C)
high_fever = temperatures[temperatures > 39.0]

percentage = (len(high_fever) / len(temperatures)) * 100

````

````python
# Find students (rows) where **all three scores are above 70**.
mask = np.all(exams > 70, axis=1)

students = exams[mask]

print("Students with all scores > 70:")
print(students)
````

````python
# Use np.where to assign grade labels: 'A' if average ≥ 80, else 'B'
averages = exams.mean(axis=1)

grades = np.where(averages >= 80, 'A', 'B')

print("Averages:", averages)
print("Grades:", grades)
````

````python
import numpy as np
````

````python
# Q3.Simulate a `(4, 3)` dataset (e.g., 4 patients, 3 measurements).   
# Manually set 3 values to `np.nan` at positions `[0,1]`, `[2,0]`, `[3,2]`

data = np.arange(1, 13, dtype=float).reshape(4, 3)

data[0,1] = np.nan
data[2,0] = np.nan
data[3,2] = np.nan
````

````python
print("Dataset:")
print(data)
````

````python
# Count NaN values **per column** (hint: `np.isnan(arr).sum(axis=0)`).  
nan_count = np.isnan(data).sum(axis=0)
print("NaN Count Per Column:")
print(nan_count)
````

````python
# Fill each column's NaN values with that column's `np.nanmean()`.

column_means = np.nanmean(data, axis=0)

print("Column Means:")
print(column_means)
````

````python
#Fill each column's NaN values with that column's `np.nanmean()`.

rows, cols = np.where(np.isnan(data))

data[rows, cols] = column_means[cols]

print("Filled Dataset:")
print(data)
````
