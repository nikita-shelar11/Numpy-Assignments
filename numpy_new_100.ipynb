```python
# 66 binary cross entropy loss
y_true = np.array([1, 0, 1, 0, 1, 0])

# Predicted probabilities
y_pred = np.array([0.9, 0.2, 0.8, 0.7, 0.3, 0.4])

print("y_true:")
print(y_true)

print("y_pred:")
print(y_pred)

# Clip predictions


# np.clip()
# Prevents log(0)

y_pred = np.clip(y_pred, 1e-7,1 - 1e-7)

print("Clipped Predictions:")
print(y_pred)

# Compute Binary Cross Entropy-

# Formula:
# -mean(y*log(p)+(1-y)*log(1-p))

loss = -np.mean( y_true * np.log(y_pred) +(1 - y_true) * np.log(1 - y_pred))
print(loss)

# clipping avoids numerical instability

```

```python
#95 verify matrix multiplication

import numpy as np

# Create random matrices

A = np.random.rand(4, 3)

# Shape = (4,3)

B = np.random.rand(3, 5)

# Shape = (3,5)

C = np.random.rand(5, 2)

# Shape = (5,2)

print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
print("Shape of C:", C.shape)

# Compute (A @ B) @ C

result1 = (A @ B) @ C

# A @ B
# (4,3) @ (3,5) → (4,5)

# Then:
# (4,5) @ (5,2) → (4,2)

print("\nResult1 Shape:", result1.shape)
# Compute A @ (B @ C)


result2 = A @ (B @ C)

# B @ C
# (3,5) @ (5,2) → (3,2)

# Then:
# (4,3) @ (3,2) → (4,2)

print("Result2 Shape:", result2.shape)

# Compute absolute difference


difference = np.abs(result1 - result2)

# np.abs()
# Converts all differences to positive values

max_difference = difference.max()

# max()
# Largest absolute difference

print("\nMaximum Absolute Difference:")
print(max_difference)

# Verify associativity


check = max_difference < 1e-10

# 1e-10
# Very tiny tolerance

print("\nAre both results equal?")
print(check)
```

```python
# 96 row wise L2 norm approaches
import time

# Create random array
arr = np.random.rand(500, 100)

# Shape:
# 500 rows
# 100 columns

print("Array Shape:", arr.shape)


# METHOD A
# Python Loop

start_a = time.perf_counter()

result_a = [
    np.sqrt((row ** 2).sum())
    for row in arr
]

end_a = time.perf_counter()

time_a = end_a - start_a

print("\nMethod A Time:", time_a)
# row ** 2
# Squares each element

# sum()
# Sum of squared values

# sqrt()
# Square root gives L2 norm
# METHOD B
# Fully Vectorized
start_b = time.perf_counter()

result_b = np.sqrt((arr ** 2).sum(axis=1))

end_b = time.perf_counter()

time_b = end_b - start_b

print("\nMethod B Time:", time_b)

# axis=1
# Row-wise sum

# Entire computation happens in optimized NumPy C code

# METHOD C
# In-place Vectorized

tmp = arr.copy()

# copy()
# Prevents modifying original array

start_c = time.perf_counter()

tmp **= 2

# In-place squaring

result_c = np.sqrt(tmp.sum(axis=1))

end_c = time.perf_counter()

time_c = end_c - start_c

print("\nMethod C Time:", time_c)

# Speedup Calculation
speedup = time_a / time_b

print("\nSpeedup of B over A:")
print(speedup)

# Verify results are similar

print("\nDo all methods match?")

print(np.allclose(result_a, result_b))

print(np.allclose(result_b, result_c))

# allclose()
# Checks numerical similarity
```

```python
# 98 reshape transpose and contiguity

import numpy as np

# Create array from 0 to 119
arr = np.arange(120)

print("Original Shape:", arr.shape)

print("Original Size:", arr.size)
# STEP A
# Reshape to (2,3,4,5)

a = arr.reshape(2, 3, 4, 5)

print("\nShape After Reshape:", a.shape)

print("C_CONTIGUOUS:", a.flags['C_CONTIGUOUS'])

print("F_CONTIGUOUS:", a.flags['F_CONTIGUOUS'])
# STEP B
# Transpose dimensions

b = a.transpose(3, 2, 1, 0)

# Axis order becomes:
# (5,4,3,2)

print("\nShape After Transpose:", b.shape)

print("C_CONTIGUOUS:", b.flags['C_CONTIGUOUS'])

print("F_CONTIGUOUS:", b.flags['F_CONTIGUOUS'])

# Transpose usually breaks contiguous memory order

# STEP C
# Force C-order contiguous array

c = np.ascontiguousarray(b)

# Creates contiguous memory layout

print("\nAfter ascontiguousarray:")

print("C_CONTIGUOUS:", c.flags['C_CONTIGUOUS'])

print("F_CONTIGUOUS:", c.flags['F_CONTIGUOUS'])


final = c.reshape(20, 6)

# (5,4,3,2)
# Total elements = 120
# Reshape to (20,6)

print("\nFinal Shape:", final.shape)
# STEP E
# Verify size remains constant
print("\nSize Verification:")

print("Original Size:", arr.size)

print("After Reshape:", a.size)

print("After Transpose:", b.size)

print("After Contiguous:", c.size)

print("Final Size:", final.size)
```

```python
# full  ml pipe line challenge

import numpy as np
# STEP 1
# Create datasete
X = np.random.rand(200, 5)

# 200 samples
# 5 features

# Create binary target
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)

# X[:,0]
# First column

# X[:,1]
# Second column

# Condition:
# if sum > 1 → class 1
# else class 0

print("Target y:\n", y)
# STEP 2
# Standardize features

mean = X.mean(axis=0)

std = X.std(axis=0)

X_scaled = (X - mean) / std

print("\nScaled Features:\n", X_scaled)

# STEP 3
# Train/Test Split

split = int(0.8 * len(X))

# 80% training
# 20% testing

X_train = X_scaled[:split]

X_test = X_scaled[split:]

y_train = y[:split]

y_test = y[split:]

print("\nTrain Shape:", X_train.shape)

print("Test Shape:", X_test.shape)

# STEP 4
# Initialize weights

W = np.random.rand(5)

b = 0.0

print("\nInitial Weights:\n", W)

# STEP 5
# Forward Pass


Z = X_train @ W + b

# Linear combination

# Sigmoid activation
pred = 1 / (1 + np.exp(-Z))

print("\nPredictions:\n", pred)

# STEP 6
# Binary Predictions


pred_labels = (pred >= 0.5).astype(int)

# Probability ≥ 0.5 → class 1

print("\nPredicted Labels:\n", pred_labels)
# STEP 7
# Compute Accuracy


accuracy = np.mean(pred_labels == y_train)

# Boolean comparison:
# True = correct
# False = wrong

print("\nTraining Accuracy:")

print(accuracy)


Z_test = X_test @ W + b

pred_test = 1 / (1 + np.exp(-Z_test))

test_labels = (pred_test >= 0.5).astype(int)

test_accuracy = np.mean(test_labels == y_test)

print("\nTest Accuracy:")

print(test_accuracy)
```
